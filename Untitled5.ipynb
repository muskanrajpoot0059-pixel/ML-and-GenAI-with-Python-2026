```python
#PART B ) EXPLORATORY DATA ANALYSIS:
#QUESTION 4 )Distribution analysis
cols = ['rainfall_mm',
        'temperature_c',
        'fertilizer_kg',
        'yield_ton_per_hectare']

for col in cols:
    plt.figure(figsize=(6,4))
    plt.hist(df[col], bins=10)
    plt.title(f'Histogram of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

1. Rainfall (rainfall_mm)
Rainfall values are distributed fairly evenly between 300 mm and 1200 mm.
No strong peak is observed, indicating rainfall is spread across different ranges.
There are no obvious outliers or extreme concentrations.
2. Temperature (temperature_c)
Temperature values are spread across the entire range with moderate variation.
Most temperature intervals have similar frequencies.
No significant skewness or extreme outliers are visible.
3. Yield (yield_ton_per_hectare)
Most yield values lie between 4 and 6 tons per hectare.
The distribution is approximately bell-shaped with a peak near 5 tons per hectare.
Extremely low and extremely high yields occur less frequently.

```python
#QUESTION 9 ) CORRELATION ANALYSIS
#A)Generate the correlation matrix for numerical features.
import seaborn as sns
import matplotlib.pyplot as plt

# Correlation Matrix
corr_matrix = df.corr(numeric_only=True)

print(corr_matrix)
```

```python
#b)create a heatmap:
# Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
#c)identify the top 3 feature most correlate with crop yield .
# Top 3 features correlated with yield
corr_with_yield = corr_matrix['yield_ton_per_hectare'].sort_values(ascending=False)

print("\nTop Features Correlated with Yield:")
print(corr_with_yield)
```

```python
#QUESTION 10 GROUP BASED ANALYSIS
#CALCULATE AVERAGE YIELD :
#.EACH CROP TYPE
crop_avg = df.groupby('crop_type')['yield_ton_per_hectare'].mean()

print("Average Yield by Crop Type:")
print(crop_avg)
```

```python
#.EACH SOIL TYPE:
# Average yield by soil type
soil_avg = df.groupby('soil_type')['yield_ton_per_hectare'].mean()

print("\nAverage Yield by Soil Type:")
print(soil_avg)
```

```python
#Which crop type and soil tyoe have highest average yield ?
print("\nCrop with Highest Average Yield:")
print(crop_avg.idxmax())

print("\nSoil Type with Highest Average Yield:")
print(soil_avg.idxmax())
```

```python
#PART C : DATA PREPARTION:
#QUESTION 11)FEATURE ENCODING;
#A)Identify the categorical columns
cat_cols = df.select_dtypes(include='object').columns

print("Categorical Columns:")
print(cat_cols)
```

```python
#B)convert them into numerical form using one-hot encoding.
df_encoded = pd.get_dummies(df, columns=cat_cols)

```

```python
#C)display the first five rows of transfered data.

print("\nFirst 5 Rows of Transformed Dataset:")
print(df_encoded.head())
```

```python
#QUESTION 12)FEATURE SELECTION;
#.separate
#input feature(X):
X = df_encoded.drop('yield_ton_per_hectare', axis=1)
```

```python
#target variable(Y):
y = df_encoded['yield_ton_per_hectare']
```

```python
print("Shape of X:", X.shape)
print("Shape of y:", y.shape)
```

```python
#PART D )MACHINE LEARNING:
#QUESTION 13 :TRAIN-TEST MODEL
#SPLIT THE DATASET INTO
#80% Training data and 20% testing data
#display the shape of each also
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)
```

```python
#QUESTION 14 : LINEAR REGRESSION MODEL:
#A)train a linear regression model
from sklearn.linear_model import LinearRegression
import pandas as pd

# Train Model
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
#B)Display the model cofficient and intercept :
# Intercept
print("Intercept:", model.intercept_)

# Coefficients
coef_df = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': model.coef_
})
print("\nCoefficients:")
print(coef_df)
```

```python
#c)which feature has the highest positive cofficient:
highest_feature = coef_df.loc[coef_df['Coefficient'].idxmax()]

print("\nFeature with Highest Positive Coefficient:")
print(highest_feature)
```
